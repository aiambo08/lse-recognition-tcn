# LSE Recognition — Modular Pipeline Quickstart

Demostración interactiva del paquete **`lse_recognition`**.
Permite entrenar, evaluar y probar el modelo TCN con solo unas pocas líneas de código limpio y modular.

```python
import sys
from pathlib import Path

# Añadir src al path si no está instalado como paquete
sys.path.insert(0, str(Path.cwd().parent / "src"))

import torch
from lse_recognition import load_config, run_training, evaluate_model, create_dataloaders
```

## 1. Cargar Configuración Externalizada (`configs/default.yaml`)

```python
config = load_config("../configs/default.yaml")
print(f"Configuración cargada:")
print(f" - Seq Length:     {config['seq_length']}")
print(f" - Input Features: {config['input_features']}")
print(f" - Num Classes:    {config['num_classes']}")
print(f" - TCN Channels:   {config['tcn_channels']}")
print(f" - Seed:           {config['seed']}")
```

## 2. Entrenamiento Completo con Pipeline Modular

```python
# Cambiamos temporalmente el directorio de trabajo para rutas relativas de datos si es necesario
import os
if Path.cwd().name == "notebooks":
    os.chdir("..")

trainer, history, test_loader, idx_to_word = run_training(
    config=config,
    use_hands_only=True,
    normalize=True,
    seed=42
)
```

## 3. Evaluación en Test Set y Matriz de Confusión

```python
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
acc, f1 = evaluate_model(
    model_path="models/best_model.pth",
    test_loader=test_loader,
    idx_to_word=idx_to_word,
    device=device,
    show_plots=True
)
```
